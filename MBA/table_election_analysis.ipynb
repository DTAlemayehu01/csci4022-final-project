```python
import pandas as pd
import numpy as np
from mlxtend.frequent_patterns import apriori, association_rules, fpmax
```

```python
# https://en.wikipedia.org/wiki/Comparison_of_electoral_systems#Compliance_of_selected_single-winner_methods
df = pd.read_csv("./voting_table_wikipedia.txt", sep='\t')
display(df.columns)
df
```

```python
df.columns = df.columns.str.replace(r'\[.*?\]', '', regex=True).str.strip()
for col in df.columns:
    if col != 'Ballot type':
        df[col] = df[col].fillna('Yes').replace('', 'Yes').replace(r'LIIA Only.*', 'Yes', regex=True).replace(r'N/A.*', 'Yes', regex=True).replace(r'Yes.*', 'Yes', regex=True)
        df[col] = df[col].replace(r'No.*', 'No', regex=True)
        df[col] = (df[col] == 'Yes').astype(int)
display(df.columns)
df
```

```python
df.eq(1).sum(axis=1)
```

```python
df.drop(['Ballot type'], axis=1)
```

```python
frequent = apriori(df.drop(['Ballot type'], axis=1), min_support=0.3, use_colnames=True)
rules = association_rules(frequent, metric='confidence', min_threshold=0.6)
maxfp = fpmax(df.drop(['Ballot type'], axis=1), min_support=0.3, use_colnames=True)
```

```python
for items in maxfp['itemsets']:
    display(items)
```

```python
pd.set_option('display.max_colwidth', 1000)
print(maxfp['itemsets'])
```

```python
print(f"Found {len(rules)} association rules")
print("\nTop 30 rules by lift:")
display(rules.nlargest(30, 'lift')[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```

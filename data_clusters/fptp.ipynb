```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from matplotlib.patches import Ellipse
from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture
```

```python
citizens, _ = make_blobs(n_samples=300, centers=5, n_features=2, cluster_std=0.8, random_state=42)
citizens.shape
```

```python
plt.scatter(citizens[:, 0], citizens[:, 1], alpha=0.6, s=20)
plt.title('Random Citizen Political Spread', fontsize=12)
plt.xlabel('X'); 
plt.ylabel('Y')
plt.grid(True, alpha=0.3)
```

```python
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10).fit(citizens)
display(kmeans.cluster_centers_)
display(kmeans.inertia_)

plt.scatter(citizens[:, 0], citizens[:, 1], alpha=0.6, s=20, c=kmeans.predict(citizens))
plt.plot(*zip(*kmeans.cluster_centers_), color='red', marker='*', markersize=12, linestyle='', label="Centroids")
plt.title('2 Candidate Fit', fontsize=12)
plt.xlabel('X'); 
plt.ylabel('Y')
plt.grid(True, alpha=0.3)
```
